```python
import src.model_preprocessing
from sklearn.linear_model import LinearRegression
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import cross_val_score
from sklearn.feature_selection import SelectKBest, VarianceThreshold, f_regression
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
import scanpy as sc
import numpy as np
import pandas as pd
from sklearn.feature_selection import RFE
```

```python

anndata = pd.read_csv("model_input_data/blood_healthy_age_scaled.csv", sep='\t')
```

```python
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
anndata = anndata.rename(columns={"Unnamed: 0": "Sample"})
anndata.set_index("Sample")
```

```python
num_cols = anndata.select_dtypes(include='number').columns

corr_matrix = anndata[num_cols].corr(method='spearman')
corr_matrix
```

```python
X = anndata.drop(columns=["Age"])
y = anndata["Age"]
```

```python
categorical_transformer = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
categorical_features = ["Sequencing_Method", "Project"]
preprocessor = ColumnTransformer(transformers=[('cat', categorical_transformer, categorical_features)])
model = LinearRegression()
rfe = RFE(
    estimator=LinearRegression(), n_features_to_select=400
)
```

```python
pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('rfe', rfe),
    ('model', model)
])
```

```python
cv = 5
scores = cross_val_score(pipeline, X, y, scoring='neg_mean_absolute_error', cv=cv)
mse_scores = -scores

mean_mse = np.mean(mse_scores)
std_mse = np.std(mse_scores)

print(f'{mean_mse = }')

print(f'{std_mse = }')
```
