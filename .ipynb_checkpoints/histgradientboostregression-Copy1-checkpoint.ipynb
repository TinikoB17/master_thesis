```python
import sklearn
import pandas as pd
import scanpy as sc
import matplotlib.pyplot as plt
import numpy as np

from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import cross_val_score, KFold
from sklearn.metrics import make_scorer, mean_absolute_error, mean_squared_error, r2_score
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import PolynomialFeatures
```

```python
annotated_obj_blood = sc.read_h5ad("filtered_input/all_human_miRNA_rpmm_harmonized_cleaned_blood.h5ad")

mirna_expression = annotated_obj_blood.to_df()
column_sums = mirna_expression.sum(axis=0, numeric_only=True)
drop_cols = column_sums[column_sums ==0].index
mirna_expression = mirna_expression.drop(columns=drop_cols)

obs_df = sc.get.obs_df(annotated_obj_blood, keys=["Age", "Disease_Condition", "Sequencing_Method"])

expr_with_age = mirna_expression.join(obs_df, how="left")
expr_with_age = expr_with_age[expr_with_age["Disease_Condition"].isin(["Healthy", "Control"])
expr_with_age
```

```python
encoder = OneHotEncoder(sparse_output=False)
categorical_cols = ["Sequencing_Method"]
expressions = expr_with_age.reset_index(names="Sample")
encoded = encoder.fit_transform(expressions[categorical_cols])

encoded_expression = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(categorical_cols))

final_expression = pd.concat([expressions.drop(categorical_cols, axis=1), encoded_expression], axis=1)
features_expressions = final_expression.set_index("Sample")
```

```python
features_expressions
```

```python
# features_expressions['Age'].value_counts()[:20].plot(kind='barh')
features_expressions = features_expressions[(features_expressions['Age'] < 90.0) & (features_expressions['Age'] > 20.0)]
features_expressions
```

```python
X = features_expressions.drop(columns=["Age", "Disease_Condition"])
y = features_expressions.drop(columns=["Disease_Condition"])["Age"]
from sklearn.feature_selection import SelectKBest, f_regression

selector = SelectKBest(score_func=f_regression, k=1000)
X = selector.fit_transform(X, y)  # Returns NumPy array

scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(X)
```

```python
# from sklearn.decomposition import PCA
# pca = PCA(n_components=150)  # keep 95% variance
# X_pca = pca.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

```python
from sklearn.ensemble import HistGradientBoostingRegressor
model = HistGradientBoostingRegressor(learning_rate=0.02,
max_iter=800,
max_leaf_nodes=127,
min_samples_leaf=5,
l2_regularization=1e-2)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
```

```python
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
```
