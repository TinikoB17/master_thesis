```python
import sklearn
import pandas as pd
import scanpy as sc
import matplotlib.pyplot as plt
import numpy as np
annotated_obj_blood = sc.read_h5ad("filtered_input/all_human_miRNA_rpmm_harmonized_cleaned_blood.h5ad")

mirna_expression = annotated_obj_blood.to_df()
column_sums = mirna_expression.sum(axis=0, numeric_only=True)
drop_cols = column_sums[column_sums < 500].index
mirna_expression = mirna_expression.drop(columns=drop_cols)

obs_df = sc.get.obs_df(annotated_obj_blood, keys=["Age", "Disease_Condition", "Sequencing_Method"])

expr_with_age = mirna_expression.join(obs_df, how="left")
```

```python
expr_with_age
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False)
categorical_cols = ["Disease_Condition", "Sequencing_Method"]
expressions = expr_with_age.reset_index(names="Sample")
encoded = encoder.fit_transform(expressions[categorical_cols])

encoded_expression = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(categorical_cols))

final_expression = pd.concat([expressions.drop(categorical_cols, axis=1), encoded_expression], axis=1)
features_expressions = final_expression.set_index("Sample")
features_expressions.columns.to_list()[-30:]
```

```python
from sklearn.feature_selection import VarianceThreshold

features_to_keep = ['Disease_Condition_AML patient',
 'Disease_Condition_Genetic Cohort',
 'Disease_Condition_C',
 'Disease_Condition_COVID-19',
 'Disease_Condition_Control',
 'Disease_Condition_Genetic Registry',
 'Disease_Condition_Healthy',
 'Disease_Condition_IgAN patient',
 'Disease_Condition_MG',
 'Disease_Condition_MNR',
 'Disease_Condition_MR',
 'Disease_Condition_OUD',
 'Disease_Condition_P',
 'Disease_Condition_PD',
 'Disease_Condition_Prodromal',
 'Disease_Condition_SDAI',
 'Disease_Condition_SWEDD',
 'Disease_Condition_patients with osteoporosis diagnosed',
 'Disease_Condition_sCJD',
 'Disease_Condition_stroke',
 'Sequencing_Method_Illumina',
 'Sequencing_Method_Illumina Genome Analyzer II',
 'Sequencing_Method_Illumina HiSeq 2000',
 'Sequencing_Method_Illumina HiSeq 2500',
 'Sequencing_Method_Illumina MiSeq',
 'Sequencing_Method_Illumina NovaSeq 6000',
 'Sequencing_Method_NextSeq 500',
]
```
