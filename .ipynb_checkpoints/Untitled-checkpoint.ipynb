```python
import sklearn
import pandas as pd
import scanpy as sc
import matplotlib.pyplot as plt
import numpy as np
annotated_obj_blood = sc.read_h5ad("filtered_input/all_human_miRNA_rpmm_harmonized_cleaned_blood.h5ad")

mirna_expression = annotated_obj_blood.to_df()
column_sums = mirna_expression.sum(axis=0, numeric_only=True)
drop_cols = column_sums[column_sums == 0].index
mirna_expression = mirna_expression.drop(columns=drop_cols)

obs_df = sc.get.obs_df(annotated_obj_blood, keys=["Age"])

expr_with_age = mirna_expression.join(obs_df, how="left")
```

```python
correlations = expr_with_age.corr()
```
