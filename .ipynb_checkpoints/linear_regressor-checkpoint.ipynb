```python
import sklearn
import pandas as pd
import scanpy as sc
import matplotlib.pyplot as plt
```

```python
annotated_obj_blood = sc.read_h5ad("filtered_input/all_human_miRNA_rpmm_harmonized_cleaned_blood.h5ad")
```

```python
mirna_expression = annotated_obj_blood.to_df()
column_sums = mirna_expression.sum(axis=0, numeric_only=True)
drop_cols = column_sums[column_sums < 500].index
mirna_expression = mirna_expression.drop(columns=drop_cols)
```

```python
mirna_expression
```

```python
obs_df = sc.get.obs_df(annotated_obj_blood, keys=["Age"])
```

```python
obs_df
```

```python
expr_with_age = mirna_expression.join(obs_df, how="left")
expr_with_age
```

```python
y = expr_with_age["Age"]
X = expr_with_age.drop(columns=["Age"])
```

```python
y, X
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import cross_val_score, KFold
from sklearn.metrics import make_scorer, mean_absolute_error, mean_squared_error, r2_score
from sklearn.ensemble import RandomForestRegressor
import numpy as np
from sklearn.linear_model import LinearRegression
regressor = make_pipeline(StandardScaler(), LinearRegression())

kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Metrics
mae_scores = -cross_val_score(regressor, X, y, cv=kf,
                              scoring=make_scorer(mean_absolute_error, greater_is_better=False))
rmse_scores = np.sqrt(-cross_val_score(regressor, X, y, cv=kf,
                                       scoring="neg_mean_squared_error"))
r2_scores = cross_val_score(regressor, X, y, cv=kf, scoring="r2")

print("MAE:", mae_scores.mean(), "+/-", mae_scores.std())
print("RMSE:", rmse_scores.mean(), "+/-", rmse_scores.std())
print("R2:", r2_scores.mean(), "+/-", r2_scores.std())
```
